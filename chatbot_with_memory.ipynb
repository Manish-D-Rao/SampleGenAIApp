```python
import os
from dotenv import load_dotenv
from langchain_ollama import ChatOllama

load_dotenv()

os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")
```

```python
model = ChatOllama(model="gemma:2b")

model
```

```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

def get_session_history(sessionId:str)-> BaseChatMessageHistory:
    if sessionId not in store:
        store[sessionId] = ChatMessageHistory()
    
    return store[sessionId]

with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable":{"session_id":"chat1"}}
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

response = with_message_history.invoke(
    [
        SystemMessage(content="u r a helpfull assistant answer all my questions please"),
        HumanMessage(
            content=(
                "hey my name is Manish"
            )
        ),
    ],
    config=config,
)
```

```python
response
```

```python
response = with_message_history.invoke(
    [
        HumanMessage(
            content=(
                "hey what is my name?"
            )
        ),
    ],
    config=config,
)
```

```python
response
```

```python
config2 = {"configurable":{"session_id":"chat2"}}

response = with_message_history.invoke(
    [
        HumanMessage(
            content=(
                "hello what is my name and who am i?"
            )
        ),
    ],
    config=config2,
)
response
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpfull assistant. Answer all the question that i send You."),
    MessagesPlaceholder(variable_name = "messages")
])

chain = prompt|model
```

```python
chain.invoke({"messages":[HumanMessage(content = "Hi My name is Manish")]})
```

```python
with_message_history = RunnableWithMessageHistory(chain, get_session_history)
```

```python
config = {"configurable":{"session_id":"chat3"}}

response = with_message_history.invoke(
    [HumanMessage(content = "Hi My name is Manish")],
    config = config
)
```
