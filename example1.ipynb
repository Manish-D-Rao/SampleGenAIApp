```python
import os
from dotenv import load_dotenv

load_dotenv()

NEXUS_API_KEY = os.getenv("NEXUS_API_KEY")
NEXUS_BASE_URL = os.getenv("NEXUS_BASE_URL")

os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model = "gpt-4.1-nano", api_key = NEXUS_API_KEY, base_url = NEXUS_BASE_URL)
print(llm)
```

```python
result = llm.invoke("What is gen ai?")
```

```python
print(result)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "U r an expert ai engineer.provide me answers based on the question i give u."),
        ("user", "{input}")
    ]
)
prompt
```

```python
chain = prompt|llm

response = chain.invoke({"input": "Tell me about langsmith and what is it usefull for?"})
print(response)
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt|llm|output_parser

response = chain.invoke({"input": "can u tell me about ur origination?"})
print(response)
```
